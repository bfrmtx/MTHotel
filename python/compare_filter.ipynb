Version using the new `atsslib` library and a path resolver that works both in a script and in a notebook.

```python
from pathlib import Path
import sys
import importlib

base_dir = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
if str(base_dir) not in sys.path:
    sys.path.insert(0, str(base_dir))

import atsslib.atsslib as atsslib_module
atsslib_module = importlib.reload(atsslib_module)

channel = atsslib_module.channel
resolve_file_path = atsslib_module.resolve_file_path
plot_ts = atsslib_module.plot_ts
fft_quick = atsslib_module.fft_quick
plot_fft = atsslib_module.plot_fft



base = resolve_file_path("filter_data/Sarıçam/run_016/084_ADU-07e_C002_THx_128Hz")
chan4x4 = channel(base)
base = resolve_file_path("filter_data/Sarıçam/run_044/084_ADU-07e_C002_THx_128Hz")
chan16x16 = channel(base)

base = resolve_file_path("filter_data/Sarıçam/run_001/084_ADU-07e_C002_THx_2048Hz")
chan1x = channel(base)
print(f"Path      : {chan4x4.path}")
print(f"Header    : {chan4x4.header}")
print(f"Cal sensor: {chan4x4.calibration['sensor']}")
print(f"Timing    : {chan4x4.timing_info()}")
print(f"Samples   : {chan4x4.get_samples()}")

print(f"Path      : {chan16x16.path}")
print(f"Header    : {chan16x16.header}")
print(f"Cal sensor: {chan16x16.calibration['sensor']}")
print(f"Timing    : {chan16x16.timing_info()}")
print(f"Samples   : {chan16x16.get_samples()}")

print(f"Path      : {chan1x.path}")
print(f"Header    : {chan1x.header}")
print(f"Cal sensor: {chan1x.calibration['sensor']}")
print(f"Timing    : {chan1x.timing_info()}")
print(f"Samples   : {chan1x.get_samples()}")

# make a list of channels for later use
channels_fil = [chan4x4, chan16x16]
```

```python
from bokeh.plotting import figure, show
plot_ts(channels_fil)
```

```python
# empty list of spectra
list_of_spectra = []
spec4x4 = fft_quick(chan4x4, 0, 128, stack_all = True)
list_of_spectra.append(spec4x4)
spec16x16 = fft_quick(chan16x16, 0, 128, stack_all = True)
list_of_spectra.append(spec16x16)
spec1x = fft_quick(chan1x, 0, 2048, stack_all = True)
list_of_spectra.append(spec1x)
labels = ["4x4", "16x16", "1x"]
plot_fft(list_of_spectra,f_min = 10, f_max = 40, labels = labels)
```
